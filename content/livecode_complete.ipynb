```python
# Select all recipes

recipes = scraped.select(".recipe")

title_durations = []

for recipe in recipes:
    title = recipe.find("p", class_="recipe-name").text.strip()
    duration = recipe.find("span", class_="recipe-cooktime")
    duration_float = float(duration.text.rstrip("min"))
    title_durations.append({title: duration_float}) # Create a Dictionary and append to Array

print(title_durations)
```

## Complete scraper that can navigate links!

```python
import requests
from bs4 import BeautifulSoup

BASE_URL = "https://recipes.lewagon.com/"
response = requests.get(BASE_URL)
html = response.content
scraped = BeautifulSoup(html, 'html.parser')

title_descriptions = []

recipes = scraped.select(".recipe")

for recipe in recipes:    
    title = recipe.find("p", class_="recipe-name").text.strip()
    duration = recipe.find("span", class_="recipe-cooktime")
    duration_float = float(duration.text.rstrip("min"))

    title_durations.append({title: duration_float})
    
print(title_durations)
    
```
